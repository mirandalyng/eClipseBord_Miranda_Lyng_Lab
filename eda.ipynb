# EDA 

In this EDA following questions are answered

* Dimensioner: hur många rader/kolumner?

* Datatyper: numeriska, kategoriska, datum, text?

* Vad representerar varje rad? (en kund, en transaktion, en dag?)

* Saknade värden (missing data) - hur mycket, och saknas det slumpmässigt eller systematiskt?

* Duplicerade rader

* Orimliga värden/outliers - negativa åldrar, orimliga datum, etc.

* Inkonsekvent formatering (t.ex. "USA", "usa", "United States" som samma sak)

* Numeriska: medelvärde, median, spridning (std/varians), min/max, distribution (histogram)

* Kategoriska: frekvenstabeller, hur många unika kategorier?
